```python
import os
from google.colab import userdata

os.environ['SERPAPI_API_KEY'] = userdata.get('SERPAPI_API_KEY')
```

```python
from serpapi import GoogleSearch
from google.colab import userdata

consulta = input("Por favor, digite os sintomas e informações do animal (ex: sintomas, animal, raça): ")

params = {
    "q": consulta,
    "api_key": userdata.get('SERPAPI_API_KEY')
}

search = GoogleSearch(params)
results = search.get_dict()

if "organic_results" in results:
    print("\nResultados da busca:")
    for i, result in enumerate(results["organic_results"]):
        titulo = result.get("title")
        snippet = result.get("snippet")
        link = result.get("link")
        if titulo and snippet and link:
            print(f"\n[{i+1}] {titulo}")
            print(f"    {snippet}")
            print(f"    Link: {link}")
else:
    print("\nNenhum resultado orgânico encontrado para a sua busca.")
    if "error" in results:
        print(f"Erro da API: {results['error']}")

print("\nFim da busca.")
```

```python
import google.generativeai as genai
from google.colab import userdata

def obter_informacoes_veterinarias(consulta):
    """
    Obtém informações veterinárias usando a API do Gemini com base na consulta fornecida.

    Args:
        consulta (str): Uma string contendo os sintomas e informações do animal
                         (ex: sintomas, animal, raça).

    Returns:
        str: A resposta gerada pelo modelo Gemini ou uma mensagem de erro.
    """
    try:
        # Configure a chave da API do Gemini
        genai.configure(api_key=userdata.get('GOOGLE_API_KEY'))

        # Carregue o modelo Gemini Pro
        model = genai.GenerativeModel('gemini-pro')

        # Crie a mensagem para o modelo
        prompt = f"Considerando os seguintes sintomas e informações sobre o animal: '{consulta}', forneça informações relevantes sobre possíveis causas, cuidados ou recomendações. Seja conciso e informativo."

        # Gere a resposta do modelo
        response = model.generate_content(prompt)

        return response.text

    except Exception as e:
        return f"Ocorreu um erro ao acessar a API do Gemini: {e}"

if __name__ == "__main__":
    consulta_usuario = input("Por favor, digite os sintomas e informações do animal (ex: sintomas, animal, raça): ")
    resultado = obter_informacoes_veterinarias(consulta_usuario)

    print("\nResposta do Gemini:")
    print(resultado)

    print("\nFim da busca.")
```
